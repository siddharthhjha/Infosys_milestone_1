```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = pd.read_csv('GlobalWeatherRepository.csv')
data.head()
```

```python
data.describe()
```

```python
data.info()
```

```python
data.groupby('location_name')['temperature_celsius'].mean().plot(kind='bar')
plt.title('Average Temperature by City')
```

```python
average_temperature = data.groupby('location_name')['temperature_celsius'].mean()
avgdict = average_temperature.to_dict()
print(avgdict)
```

```python
countryhighlow = data.groupby('location_name')['temperature_celsius'].agg(['min', 'max'])
print(countryhighlow)
```
